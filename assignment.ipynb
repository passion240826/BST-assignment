# 3번
1. 5, 3, 3, 8, 7, 6, 9
2. 3, 3, 5, 6, 7, 8, 9
3. 3, 3, 6, 7, 8, 7, 5
4. 증명:  
- i 베이스 케이스
    > 노드가 0개 또는 1개인 이진 탐색 트리의 경우, 방문 순서는 그 자체로 정렬되어있다. 
      
- ii 귀납적 가정
    > 노드의 개수가 k개 미만인 모든 이진 탐색 트리에 대해 중위 순회가 정렬된 결과를 반환한다고 가정한다. 

- iii 귀납적 전개  
    -  노드수가 n개인 트리 T가 있다고 가정.
    - 중위 순회는 왼쪽 서브트리 --> 루트노드 --> 오른쪽 서브트리 순서로 순회한다.
    - 귀납적 과정에 의해 왼쪽 서브트리의 순회 결과는 정렬상태이고, 모든 값은 루트 노드보다 작거나 같다.
    - 오른쪽 서브트리의 순회 결과는 정렬 상태이고, 모든 값은 루트노드보다 크거나 같다.
    - 따라서 (정렬된 작은 값들) -> 루트 -> (정렬된 큰 값들)의 형태가 되어 항상 정렬된 상태로 순회하게 된다.
